Program 7:You are comparing the average daily sales between two stores. Store A has a mean daily sales value of$1,000 with a standard deviation of $100 over 30 days, and Store B has a mean daily sales value of $950with a standard deviation of $120 over 30 days. Conduct a two-sample t-test to determine if there is a significant difference between the average sales of the two stores at the 5% significance level.

```python
import math
from scipy.stats import t
```

```python
# Store A
mean_A = 1000
std_A = 100
n_A = 30
```

```python
mean_B = 950
std_B = 120
n_B = 30
```

```python
# Calculate standard error
SE = math.sqrt((std_A**2 / n_A) + (std_B**2 / n_B))

```

```python
# Calculate t-statistic
t_value = (mean_A - mean_B) / SE
```

```python
# Degrees of freedom
df = n_A + n_B - 2
```

```python
# Two-tailed p-value
p_value = 2 * (1 - t.cdf(abs(t_value), df))
```

```python
print("t-value =", round(t_value, 3))
print("p-value =", round(p_value, 4))
```

```python
# Decision
if p_value < 0.05:
    print("Reject H0: There is a significant difference in average sales.")
else:
    print("Fail to reject H0: There is no significant difference in average sales.")
```

Conclusion
H₀: Average sales of Store A = Store B
H₁: Average sales of Store A ≠ Store B
Significance level: α = 0.05
p-value ≈ 0.0825
Since 0.0825 > 0.05, we fail to reject H₀.

Therefore, there is no statistically significant difference between the average daily sales of Store A and Store B at the 5% significance level.
