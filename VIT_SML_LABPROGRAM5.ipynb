A researcher conducts an experiment with a sample of 20 participants to determine if a new drug affects heart rate. The sample has a mean heart rate increase of 8 beats per minute and a standard deviation of 2beats per minute. Perform a hypothesis test using the t-distribution to determine if the mean heart rate increase is significantly different from zero at the 5% significance level.

```python
import scipy.stats as stats
import math
```

```python
# Get values from the user at runtime
n = int(input("Enter sample size: "))
sample_mean = float(input("Enter sample mean: "))
sample_std = float(input("Enter sample standard deviation: "))
mu0 = float(input("Enter hypothesized population mean: "))
alpha = float(input("Enter significance level (e.g., 0.05): "))
```

```python
# Calculate t-statistic
t_statistic = (sample_mean - mu0) / (sample_std / math.sqrt(n))
```

```python
# Degrees of freedom
df = n - 1
```

```python
# Two-tailed p-value
p_value = 2 * stats.t.sf(abs(t_statistic), df)
```

```python
# Critical t-value
t_critical = stats.t.ppf(1 - alpha / 2, df)
```

```python
# Display results
print("\n----- One-Sample t-Test -----")
print(f"t-statistic: {t_statistic:.4f}")
print(f"Degrees of freedom: {df}")
print(f"Critical t-value: ±{t_critical:.4f}")
print(f"p-value: {p_value:.6f}")
```

```python
# Decision
if p_value < alpha:
    print("\nDecision: Reject the Null Hypothesis (H0)")
    print("Conclusion: The mean is significantly different from the hypothesized value.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H0)")
    print("Conclusion: The mean is not significantly different from the hypothesized value.")
```
