```python
df.describe()
### 2. `df.describe()`

#**Explanation:** The descriptive statistics summarize the distribution, central tendency, and spread of each numerical feature in the dataset.
```

```python
import matplotlib.pyplot as plt

# Plot histograms for all variables
df.hist(figsize=(10, 10))
plt.tight_layout()
plt.show()

## 3. Histogram (`df.hist()`)

#**Explanation:** The histograms show the distribution of each feature, helping identify skewness, spread, and potential outliers.
```

```python
import seaborn as sns
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()

### 4. Correlation Heatmap (`sns.heatmap()`)

#**Explanation:** The heatmap displays the correlation between features, indicating the strength and direction of their relationships.

```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(x='Diagnosis', y='Glucose', data=df)
plt.title('Glucose levels by Diagnosis')
plt.show()

### 5. Box Plot (`Diagnosis` vs `Glucose`)

#**Explanation:** The box plot compares glucose level distributions for diabetic and non-diabetic patients and highlights possible outliers.

```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Glucose', y='BMI', hue='Diagnosis', data=df)
plt.title('Glucose vs BMI')
plt.xlabel('Glucose')
plt.ylabel('BMI')
plt.show()

### 6. Scatter Plot (`Glucose` vs `BMI`)

#**Explanation:** The scatter plot shows the relationship between glucose and BMI while distinguishing diabetic and non-diabetic cases.

```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x='Diagnosis', y='Glucose', data=df)
plt.title('Average Glucose Level by Diagnosis')
plt.xlabel('Diagnosis')
plt.ylabel('Average Glucose Level')
plt.show()


### 7. Bar Plot (`Diagnosis` vs `Glucose`)

#**Explanation:** The bar plot compares the average glucose levels of diabetic and non-diabetic patients.
```
