A company is testing two versions of a webpage (A and B) to determine which version leads to more sales. Version A was shown to 1,000 users and resulted in 120 sales. Version B was shown to 1,200 users and resulted in 150 sales. Perform an A/B test to determine if there is a statistically significant difference in the conversion rates between the two versions. Use a 5% significance level.

# A/B Testing: Comparing Conversion Rates of Webpage A and B

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

# Enter the data  
Version A
Total users = 1000
Sales = 120
Version B
Total users = 1200
Sales = 150

```python
n_A, x_A = 1000, 120
n_B, x_B = 1200, 150
```

# Calculate conversion rates

```python
p_A = x_A / n_A
p_B = x_B / n_B
```

# Print conversion rates

```python
print("Conversion Rate A:", p_A)
print("Conversion Rate B:", p_B)
```

| Webpage | Users | Sales | Conversion Rate |
| ------- | ----: | ----: | --------------: |
| A       |  1000 |   120 |             12% |
| B       |  1200 |   150 |           12.5% |


Although Version B has a slightly higher conversion rate, we need to determine whether this difference is statistically significant.

# State the hypotheses

State the hypotheses
Null Hypothesis \(H_0\)
$$ H_0:p_A=p_B $$

There is no significant difference between the conversion rates.

Alternative Hypothesis \(H_1\)
$$ H_1:p_A\neq p_B $$

There is a significant difference between the conversion rates.

Since we are checking for any difference, we use a two-sided test.

# Create arrays for the Z-test

```python
count = np.array([x_A, x_B])
```

```python
count
```

Here:
120 = sales from Version A
150 = sales from Version B

```python
nobs = np.array([n_A, n_B])
```

```python
nobs
```

Here:

1000 = users who saw Version A
1200 = users who saw Version B

# Perform the two-proportion Z-test

```python
z_stat, p_value = proportions_ztest(
    count,
    nobs,
    alternative='two-sided'
)
```

# Print the results

```python
print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

# Set the significance level

```python
alpha = 0.05
```

```python
if p_value < alpha: 
    print("Reject H0: Significant difference in conversion rates.") 
else:
    print("Fail to reject H0: No significant difference in conversion rates.")
```

Final Interpretation
Conversion Rates
Version A = 12%
Version B = 12.5%
Statistical Test
Z-statistic ≈ -0.36
P-value ≈ 0.72
Significance level = 0.05

Since:

$$ \boxed{p\text{-value}>0.05} $$

we fail to reject the null hypothesis.

Conclusion

There is no statistically significant difference between the conversion rates of Webpage A and Webpage B at the 5% significance level.

Although Version B has a slightly higher observed conversion rate (12.5% vs. 12%), the observed difference of 0.5 percentage points could reasonably be due to random sampling variation.

```python
Webpage A → 12% conversion
Webpage B → 12.5% conversion

Difference → 0.5%

But...

P-value = 0.72 > 0.05

Therefore:
Fail to reject H₀

Conclusion:
No statistically significant difference
```

Practical takeaway: Based on this A/B test alone, there is insufficient statistical evidence to conclude that Version B performs better than Version A.
