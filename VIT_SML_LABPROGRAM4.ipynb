# Program 4
You want to estimate the mean salary of software engineers in a country. You take 10 different random samples, each containing 50 engineers, and calculate the sample mean for each. Plot the distribution of these sample means. How does the Central Limit Theorem explain the shape of this sampling distribution, even if the underlying salary distribution is skewed?

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set random seed for reproducibility
np.random.seed(42)
```

```python

# Create a skewed population using log-normal distribution
population_size = 10000
population = np.random.lognormal(
    mean=10,
    sigma=0.5,
    size=population_size
)
```

```python
# Take 1000 random samples of 50 engineers
n_samples = 1000
sample_size = 50
sample_means = []
```

```python
for i in range(n_samples):
    sample = np.random.choice(population,size=sample_size,replace=False)
    sample_means.append(np.mean(sample))
```

```python
# Plot histograms
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Histogram of original population
ax1.hist(population, bins=50, edgecolor='black')
ax1.set_title("Original Population")
ax1.set_xlabel("Salary")
ax1.set_ylabel("Frequency")

# Histogram of sample means
ax2.hist(sample_means, bins=30, edgecolor='black')
ax2.set_title("Sampling Distribution of Sample Means")
ax2.set_xlabel("Sample Mean")
ax2.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

What you should see

Histogram 1: Original Population

It will be right-skewed.
Salaries will have a long tail toward higher values.

Histogram 2: Sample Means

It should look much more bell-shaped/approximately normal.
The values will be concentrated around approximately $24,958.
It will have considerably less spread than the original population.

```python
# Print results
print(f"Population mean: ${np.mean(population):,.2f}")
print(f"Population standard deviation: ${np.std(population):,.2f}")
print(f"Mean of sample means: ${np.mean(sample_means):,.2f}")
print(f"Standard deviation of sample means: ${np.std(sample_means):,.2f}")
```

Observation: The histogram of the original population shows a skewed distribution, whereas the histogram of the sample means is approximately bell-shaped and centered around the population mean. The mean of the sample means ($24,958.31) is very close to the population mean ($24,956.58). Also, the standard deviation of the sample means ($1,902.30) is much smaller than the population standard deviation ($13,383.15). This demonstrates the Central Limit Theorem and the reduced variability of the sampling distribution.
