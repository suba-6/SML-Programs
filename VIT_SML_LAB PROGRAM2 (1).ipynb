# Lab Program 2:You are given a dataset with categorical variables about customer satisfaction levels (Low, Medium, High) and whether customers made repeat purchases (Yes/No). Create visualizations such as bar plots or stacked bar charts to explore the relationship between satisfaction level and repeat purchases. What can you infer from the data?

```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create a dummy dataset
data = {
    'Satisfaction Level': np.random.choice(
        ['Low', 'Medium', 'High'],
        size=200
    ),
    'Repeat Purchase': np.random.choice(
        ['Yes', 'No'],
        size=200,
        p=[0.4, 0.6]   # 40% Yes, 60% No
    )
}
```

```python
# Fixed dataset
data = {
    'Satisfaction Level': (
        ['Low'] * 40 +
        ['Medium'] * 40 +
        ['High'] * 40 +
        ['Low'] * 10 +
        ['Medium'] * 20 +
        ['High'] * 30
    ),
    'Repeat Purchase': (
        ['No'] * 40 +
        ['No'] * 40 +
        ['No'] * 40 +
        ['Yes'] * 10 +
        ['Yes'] * 20 +
        ['Yes'] * 30
    )
}
```

```python
# Create DataFrame
df_customer = pd.DataFrame(data)
```

```python
# Display first 5 rows
print(df_customer.head())
```

```python
# Create a contingency table (cross-tabulation)
cross_tab = pd.crosstab(
    df_customer['Satisfaction Level'],
    df_customer['Repeat Purchase']
)

print("\nContingency Table:")
print(cross_tab)
```

```python
# Plot stacked bar chart
cross_tab.plot(
    kind='bar',
    stacked=True,
    figsize=(8, 6),
    color=['black', 'purple']
)

plt.title('Repeat Purchase by Satisfaction Level')
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.legend(title='Repeat Purchase')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Create Grouped Bar Chart
cross_tab.plot(
    kind='bar',
    stacked=False,
    figsize=(8,6),
    color=['black', 'purple']
)

plt.title('Repeat Purchase by Satisfaction Level')
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.legend(title='Repeat Purchase')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

Inference
High Satisfaction: 30 customers made repeat purchases, indicating high customer loyalty.
Medium Satisfaction: 20 customers made repeat purchases, indicating moderate loyalty.
Low Satisfaction: Only 10 customers made repeat purchases, indicating low loyalty.
The chart shows a positive relationship between customer satisfaction and repeat purchases: as satisfaction increases, the number of repeat purchases also increases.
