Program 10 :A hospital is using a Poisson regression model (a type of GLM) to predict the number of emergency room visits per week based on patient age and medical history. The model is given by: Log(λ) =2.5-0.03*Age+0.5*condition where λ is the expected number of visits per week, Age is the patient's age, and condition is a binary variable (1 if the patient has a chronic condition, 0 otherwise). Interpret the coefficients of Age and condition. What is the expected number of visits per week for a 60-year-old patient with a chronic condition? How would the expected number of visits change if the patient did not have a chronic condition? 

```python
import numpy as np

# Given coefficients
intercept = 2.5
age_coefficient = -0.03
condition_coefficient = 0.5

# Patient details
age = 60

# Case 1: Patient has chronic condition
condition = 1

log_lambda = (intercept
              + age_coefficient * age
              + condition_coefficient * condition)

lambda_value = np.exp(log_lambda)

print("Patient: 60 years old with chronic condition")
print("Expected visits per week:", round(lambda_value, 2))


# Case 2: Patient does not have chronic condition
condition = 0

log_lambda = (intercept
              + age_coefficient * age
              + condition_coefficient * condition)

lambda_value_no_condition = np.exp(log_lambda)

print("\nPatient: 60 years old without chronic condition")
print("Expected visits per week:", round(lambda_value_no_condition, 2))


# Interpret coefficients
print("\nInterpretation:")
print("Age coefficient = -0.03")
print("Each additional year of age decreases expected visits by about 2.96%.")

print("\nCondition coefficient = 0.5")
print("A chronic condition increases expected visits by about 64.87%.")
```
